```python
# W5A1
def Max_Of_Two(a, b):
    return a if a > b else b
result = Max_Of_Two(5, 10)
print(result) 
```

```python
# W5A3
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
print(is_prime(8))
```

```python
# W5A4
def is_perfect(n):
    divisors_uoc = [i for i in range(1, n) if n % i == 0]
    return sum(divisors_uoc) == n
print(is_perfect(6))
```

```python
# W5A5
def find_greater_than_k(lst, k):
    for i, num in enumerate(lst):
        if num > k:
            return i
    return -1
lst = list(map(int, input().split()))
k = int(input())
print(find_greater_than_k(lst, k))
```

```python
# W5A6
def factorial(n):
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
n = int(input())
print(factorial(n))
```

```python
# W5AA7
def hamming_distance(x, y):
    return bin(x ^ y).count('1')
x, y = map(int, input().split())
print(hamming_distance(x, y))
```

```python
# W5A8
def hamming_distance(str1, str2):
    # Đảm bảo cả hai chuỗi có độ dài bằng nhau
    if len(str1) != len(str2):
        raise ValueError("Hai chuỗi phải có độ dài bằng nhau")
    
    # Tính số lượng ký tự khác nhau tại mỗi vị trí
    distance = sum(1 for a, b in zip(str1, str2) if a != b)
    return distance
str1 = input()
str2 = input()
print(hamming_distance(str1, str2))
```

```python
# W5A9
def sum_digits(n):
    return sum(int(digit) for digit in str(n))
n = int(input())
print(sum_digits(n))
```
