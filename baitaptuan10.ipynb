```python
# W9A1
def merge_lists(list1, list2):
    ket_qua = []
    i = 0 
    j = 0 

    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:  
            ket_qua.append(list1[i]) 
            i += 1                
        else:                     
            ket_qua.append(list2[j]) 
            j += 1               
    
    
    ket_qua.extend(list1[i:])
    ket_qua.extend(list2[j:])
    
    return ket_qua


print(merge_lists([1, 2, 4], [1, 3, 4]))
```
