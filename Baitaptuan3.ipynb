```python
#W3A16
x = float(input())
phan_nguyen = int(x) # Cắt bỏ phần thập phân -> Tròn xuống

# Tròn lên: Nếu x > phần nguyên (tức là có lẻ) thì cộng 1
tron_len = phan_nguyen + 1 if x > phan_nguyen else phan_nguyen

# Tròn gần nhất: Cộng 0.5 rồi lấy phần nguyên
tron_gan_nhat = int(x + 0.5)
print("Phần nguyên", phan_nguyen ,", Tròn lên:", tron_len, ", Tròn gần nhất:", tron_gan_nhat)
```

```python
#W3A17
a, b, c, d = map(int, input("Nhập 4 số a, b, c, d").split())
if a==0:
    print("Không là Cấp số nhân")
elif b % a == 0:
    q = b // a
    if c == b * q and d == c * q:
        print("Là Cấp số nhân")
    else:
        print("Không là Cấp số nhân")
else:
    print("Không là Cấp số nhân")
```
