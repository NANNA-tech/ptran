```python
#Bài 3
import math
a, b, c = map(float, input("Nhập cạnh a, b, c: ").split())

if a + b > c and a + c > b and b + c > a:
    print("Là tam giác.")
    p = (a+b+c)/2
    print(f"Chu vi là: {p*2}, Diện tích là: {math.sqrt(p*(p-a)*(p-c)*(p-b))}")
    if a==b==c:
        print("Tam giác đều.")
    elif a == b or b == c or a ==c:
        print("Tam giác cân.")
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a:  
        print("Tam giác vuông.")  
    else:
        print("Tam giác thường.")     
else:
    print("Khong la ba canh cua mot tam giac!")
```

```python
#Bài 5:(ax^2 + bx + c = 0)
a, b, c = map(float, input("Nhập a, b: ").split())
delta=b**2 - 4*a*c
if delta < 0:
    print("Phương trình vô nghiệm.")
elif delta == 0:
    print(f"Phương trình có nghiệm kép: x = {-b/(2*a)}")   
else:
    x1 = (- b + math.sqrt(delta))/(2*a)
    x2 = (- b - math.sqrt(delta))/(2*a)
    print(f"Phương trình có 2 nghiệm phân biệt: x1 = {x1}, x2 = {x2}")
```

```python
#Bài 8
def giai_he_phuong_trinh(a, b, m,c,d, n):
    det = a * d - b * c
    if det == 0:
        return "Hệ phương trình vô nghiệm hoặc vô số nghiệm"
    else:
        x = (m * d - b * n) / det
        y = (a * n - m * c) / det
        return f"Giải: x = {x}, y = {y}"
a, b, m, c, d, n = map(float, input().split())
print(giai_he_phuong_trinh(a, b, m, c, d, n))
```

```python
#Bài 9
n = int(input("Nhập giây: "))
Phut = n//60
Gio = Phut//60
Phutmoi = max(0, Phut - Gio*60)
Giaymoi = max(0, n - Gio*3600 - Phutmoi*60)
print(Gio, "giờ", Phutmoi, "phút", Giaymoi, "giây")
```

```python
#Bài 10
import math
tam_x = float(input("Nhập hoành độ tâm: "))
tam_y = float(input("Nhập tung độ tâm: "))
r = float(input("Nhập bán kính: "))

#Nhập điểm A
x = float(input("Nhập hoành độ A(x): "))
y = float(input("Nhập tung độ A(y): "))

khoang_cach_binh_phuong = (x - tam_x)**2 + (y - tam_y)**2

if khoang_cach_binh_phuong <= r**2:
    print("Điểm A thuộc đường tròn.")
else:
    print("Điểm A nằm ngoài đường tròn.")   
```
