```python
# cach 2  (ĐẢM VẢO CHỈ TÁCH ĐÔI, ĐỘ DÀI CỦA MỖI CẶP LÀ PHẢI ĐÚNG LÀ 2 THÌ MỚI XÉT, CHẶT CHẼ HƠN TRONG TRƯỜNG HỢP USER NHẬP SAI YÊU CẦU)
pairs = input().split()   #split ở đây là cắt, cứ thấy dấu cắt là cách
result = {}
for p in pairs:
    data = p.split(":", 1)
    if len(data) == 2:
        key = data[0]
        value = data[1]
        if key not in result:
            result[key] = []
        result[key].append(value)
print(result)
```

```python
# W6A5
arr = list(map(int, input().split()))
result = {'positives': 0, 'negatives': 0, 'zeros': 0}

for x in arr:
    if x > 0:
        result['positives'] += 1
    elif x < 0:
        result['negatives'] += 1
    else:
        result['zeros'] += 1

print(result)
```

```python
#A9
d1 = eval(input("NHẬP ĐÚNG DẠNG DICTIONARY:"))
d2 = eval(input("NHẬP ĐÚNG DẠNG DICTIONARY:"))

result = {}

for k in d1:
    result[k] = d1[k]

for k in d2:
    if k in result:
        result[k] += d2[k]
    else:
        result[k] = d2[k]

result = dict(sorted(result.items()))
print(result)
```

```python
#A10
arr = list(map(int, input().split()))
k = int(input())

pairs = []

n = len(arr)
for i in range(n):
    for j in range(i + 1, n):
        if arr[i] + arr[j] == k:
            pairs.append((arr[i], arr[j]))

pairs.sort(key=lambda x: x[0])
print(pairs)
```

```python
#A11
t = tuple(map(int, input().split()))

even = []
odd = []

for x in t:
    if x % 2 == 0:
        even.append(x)
    else:
        odd.append(x)

print(tuple(even))
print(tuple(odd))
```

```python
#A12
arr = list(map(int, input().split()))
count = {}

for x in arr:
    if x not in count:
        count[x] = 0
    count[x] += 1

max_freq = max(count.values())
candidates = [x for x in count if count[x] == max_freq]

print(min(candidates))
```

```python
#A14
a = list(map(int, input().split()))
b = list(map(int, input().split()))

result = []
seen = set()

for x in a:
    if x in b and x not in seen:
        result.append(x)
        seen.add(x)

print(result)
```
