```python
# W4A2
while True:
    n = int(input("Nhập số nguyên dương n: "))
    if n > 0:
        break
# Kiểm tra nguyên tố
is_prime = True
if n < 2:
    is_prime = False
else:
    # Chạy từ 2 đến căn bậc 2 của n
    i = 2
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i += 1

if is_prime:
    print(f"{n} là số nguyên tố")
else:
    print(f"{n} KHÔNG là số nguyên tố")
```

```python
# W4A6
a = int(input())
b = int(input())
tong = 0

for num in range(a, b + 1):
    # Kiểm tra num có phải nguyên tố không
    if num < 2:
        continue
    is_p = True
    j = 2
    while j * j <= num:
        if num % j == 0:
            is_p = False
            break
        j += 1
    
    if is_p:
        tong += num
print(tong)
```

```python
# W4A7
n = int(input("Nhập n (n >= 2): "))
max_prime = -1
d = 2

# Thuật toán phân tích thừa số nguyên tố
while d * d <= n:
    while n % d == 0:
        max_prime = d
        n //= d
    d += 1
if n > 1:
    max_prime = n # Phần còn lại chính là ước nguyên tố lớn nhất

print(max_prime)
```

```python
# W4A8
n = int(input("Nhập n: "))
steps = 0

while True:
    # 1. Tìm số đảo ngược của n
    temp = n
    rev_n = 0
    while temp > 0:
        rev_n = rev_n * 10 + temp % 10
        temp //= 10
    
    # Kiểm tra n có phải Palindrome chưa (n == rev_n)
    if n == rev_n:
        break # Dừng vòng lặp
    
    # Nếu chưa, cộng n với số đảo ngược
    n = n + rev_n
    steps += 1

print(f"{steps} {n}")
```

```python
# WW4A9
n = int(input("Nhập n: "))
found = False

i = 1
while i * i <= n:
    sq = i * i
    
    # Kiểm tra các chữ số của sq có khác nhau không
    temp = sq
    is_distinct = True
    # Dùng 2 vòng lặp lồng nhau để so sánh các chữ số (vì ko được dùng mảng/chuỗi)
    # Cách đơn giản hơn: Dùng bitmask (nâng cao) hoặc tách số
    # Ở đây mình dùng cách tách số thủ công:
    
    mask = 0 # Dùng 1 số nguyên làm bộ nhớ (bitmask)
    temp_check = sq
    while temp_check > 0:
        digit = temp_check % 10
        if (mask >> digit) & 1: # Nếu bit thứ digit đã bật -> trùng
            is_distinct = False
            break
        mask = mask | (1 << digit) # Bật bit thứ digit lên
        temp_check //= 10
        
    if is_distinct:
        print(sq, end=" ")
        found = True
    i += 1

if not found:
    print("no number")
else:
    print() # Xuống dòng
```

```python
# W4A10
n = int(input("Nhập n: "))
max_len = 0
best_num = 1

for x in range(1, n + 1):
    curr = x
    length = 1
    while curr != 1:
        if curr % 2 == 0:
            curr //= 2
        else:
            curr = 3 * curr + 1
        length += 1
    
    if length > max_len:
        max_len = length
        best_num = x
    # Nếu bằng nhau thì giữ x nhỏ hơn (do x chạy từ 1 lên nên tự thỏa mãn)

print(f"{best_num} {max_len}")
```

```python
# WA11
n = int(input("Nhập n: "))
count = 0
for i in range(1, int(n**0.5) + 1):
    if n % i == 0:
        if i % 2 == 0:
            count += 1
        j = n // i
        if j != i and j % 2 == 0:
            count += 1
print(count)
```

```python
# W4A13
line = input("Nhập 2 số cách nhau bởi dấu cách: ").split()
if len(line) == 2:
    num1, num2 = map(int, line)
    
    # Tính tổng ước của num1 (trừ chính nó)
    sum1 = 0
    for i in range(1, num1): # Chạy từ 1 đến num1-1
        if num1 % i == 0:
            sum1 += i
            
    # Tính tổng ước của num2 (trừ chính nó)
    sum2 = 0
    for i in range(1, num2):
        if num2 % i == 0:
            sum2 += i
            
    if sum1 == num2 and sum2 == num1:
        print("true")
    else:
        print("false")
```

```python
# W4A15
tong_con = int(input("Tổng số con: "))
tong_chan = int(input("Tổng số chân: "))
found_sol = False

# Gà có 2 chân, chó có 4 chân
# Giả sử số gà là g, chạy từ 0 đến tong_con
for ga in range(tong_con + 1):
    cho = tong_con - ga
    if ga * 2 + cho * 4 == tong_chan:
        print(f"{ga} {cho}")
        found_sol = True
        break # Tìm thấy 1 nghiệm là dừng (hoặc in hết tùy đề)

if not found_sol:
    print("invalid")
```
