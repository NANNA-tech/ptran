```python
# XÁC ĐỊNH TAM GIÁC
a,b,c=map(float, input().split())

if a+b>c and a+c>b and b+c>a:
  if a==b==c:
   print("Tam giác đều")
  elif (a==b and a!=c) or (a==c and a!=b) or (b==c and b!=a):
   print("Tam giác cân")
  else:
    print("Tam giác thường")
else:
   print("Không phải tam giác")
```

```python
# viết * tạo thành tam giác cân
h = int(input())
for i in range(1, h + 1):
    print(" " * (h - i), end="")
    if i == 1:
        print("*")
    elif i == h:
        print("*" * (h*2 -1))
    else:
        print("*" + " " * (2 * i - 3) + "*")
```

```python
import math
def check_perfect_numbers(x : int) -> bool:   # bool chữ kí của hàm : True/False
      tong_uoc = 0
      for i in range(1, math.isqrt(x) + 1):    
            if x % i == 0:    # (variable i: int)
                  tong_uoc += 1
                  tong_uoc += x//i
      if(math.isqrt(x)**2):
            tong_uoc -= math.isqrt(x)
      return tong_uoc == x




N = int(input())
for x in range(2, N + 1):
      if(check_perfect_numbers(x)):
            print(x, end = ' ')
```

```python
def check_perfect_numbers(x : int) -> bool:   # bool chữ kí của hàm : True/False
      tong_uoc = 0
      for i in range(1, x//2 +1):    
            if x % i == 0:    # (variable i: int)
                  tong_uoc += i
                  
      return tong_uoc == x

N = int(input())
for x in range(2, N + 1):
      if(check_perfect_numbers(x)):
            print(x, end = ' ')
```

```python
def sumEvenDigits(x : int):
    sumEvenDigit = 0
    while x > 0:
        donVi = x % 10
        if (donVi % 2 == 0):
            sumEvenDigit += donVi
        x //= 10
    return sumEvenDigit
Number = int(input())
print(sumEvenDigits(Number))
```
