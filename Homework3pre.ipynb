```python
# 4
h = int(input())    # viet theo chieu tu tren xuong duoi
for i in range(1, h + 1):
    print(" " * (h - i), end= "")   # số khoảng trắng ở mỗi đầu dòng giảm dần khi xuống dòng
                                    # CẦN (end="") vì nếu không có thì laptop in print(" ") xong lập tức xuống dòng mới. Khi lệnh in sao chạy, nó sẽ in ở đầu dòng tiếp theo (sát lề trái)
                                    # end= "" để đẩy ngôi sao ra chính giữa, end="" giúp giữ con chỏ chuột ở lại trên cùng 1 dòng sau khi in xong.
    if i == 1:   # dinh tam giac chi in 1 *
        print("*")
    elif i == h: # day tam giac, neu h = 4 thi se can in bao nhieu ngoi sao de tao thanh tam giac can nhu sample
        print("*" * (2 * h - 1))
    else: # cac hang o giua: in kieu sao + khoang trang + sao
        print("*" + " " * (2 * i - 3) + "*")  # cong thuc khoang trang o giua: vi du o hang i = 3: can (2*3 - 3 = 3) 3 khoang trang o giua (**). hang i = 2 can 1 khoang trang o giua (2*2 - 3)
```

```python
h = int(input())

for i in range(1, h + 1):
    # Dòng đầu tiên (Đỉnh tam giác)
    if i == 1:
        print(" " * (h - 1) + "*")
        
    # Dòng cuối cùng (Đáy tam giác)
    elif i == h:
        print("*" * (2 * h - 1))
        
    # Các dòng ở giữa (Thân tam giác rỗng)
    else:
        # Công thức: Khoảng trắng lề trái + Sao trái + Khoảng trắng ở giữa + Sao phải
        print(" " * (h - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
# 14
so_dau = int(input())
if so_dau == -1:
    print("Bạn chưa nhập số nào cả")
else:
    max_number = so_dau
    min_number = so_dau
    while True:
        b = int(input())
        if b == -1:
            break
        if b > max_number:
            max_number = b
        if b < min_number:
            min_number = b
    print(f"max = {max_number}")
    print(f"min = {min_number}")
```

```python
# 16
n = int(input())
temp = n        # Tạo bản sao để xử lý
chan = 0        # Biến đếm số chẵn (ban đầu chưa có gì)
le = 0          # Biến đếm số lẻ

if n == 0:      # Xử lý riêng trường hợp số 0 (vì 0 là số chẵn)
    chan = 1
else:
    # Nếu n âm, ta lấy trị tuyệt đối để tính toán cho dễ
    temp = abs(n) 
    
    while temp > 0:
        digit = temp % 10  # BƯỚC 1: Lấy chữ số cuối cùng
        
        # BƯỚC 2: Kiểm tra chẵn lẻ
        if digit % 2 == 0: # Nếu chia hết cho 2 là chẵn
            chan += 1
        else:              # Ngược lại là lẻ
            le += 1
            
        temp = temp // 10  # BƯỚC 3: Cắt bỏ chữ số vừa xét, TIẾP TỤC ĐẾN KHI HẾT SỐ
        
print(f"Chẵn: {chan}, Lẻ: {le}")
print(f"Số chữ số chẵn: {chan}, Số chữ số lẻ: {le}")
```

```python
# 18
n = int(input("Nhập n: "))
original_n = n  # Lưu lại số gốc để lát in ra cho đẹp

# 1. Loại trừ các trường hợp sai ngay từ đầu
if n < 1:
    print("Sai")
else:
    # 2. Vòng lặp "Lọc số 3"
    # Chừng nào n còn chia hết cho 3, thì cứ chia tiếp
    while n % 3 == 0:
        n = n // 3  # Chia n cho 3 (cập nhật lại n)

    # 3. Kiểm tra tàn dư cuối cùng
    if n == 1:
        print(f"{original_n} là số dạng 3^k")
    else:
        print(f"{original_n} KHÔNG phải dạng 3^k")
```

```python
# 19
A = int(input("Nhập giới hạn A: "))

# Hai số đầu tiên của dãy
f1 = 1
f2 = 1

if A < 1:
    print("Không có số Fibonacci nào")
else:
    # In ra 2 số đầu tiên trước (vì vòng lặp tính từ số thứ 3)
    print(f1, end=" ")
    count = 1
    
    if A >= 1: # Xử lý trường hợp số 1 thứ hai
         print(f2, end=" ")
         count += 1
    
    while True:
        f_next = f1 + f2
        if f_next > A:
            break # Dừng nếu vượt quá A
            
        print(f_next, end=" ")
        count += 1
        
        # Cập nhật số để tính vòng tiếp theo
        f1 = f2
        f2 = f_next

    print(f"\nTổng cộng có {count} số Fibonacci không vượt quá {A}")
```

```python
# 20
n = int(input("Nhập n: "))
temp = n
count_nt = 0
while temp > 0:
    digit = temp % 10
    # Kiểm tra thủ công vì chỉ có 4 số
    if digit == 2 or digit == 3 or digit == 5 or digit == 7:
        count_nt += 1
    temp = temp // 10
print(f"Số lượng chữ số nguyên tố là: {count_nt}")
```
