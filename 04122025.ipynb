```python
# KIỂM TRA XEM LÀ SỐ NGUYÊN TỐ HAY KHÔNG?
import math
n = int(input("Nhập số nguyên "))
if n > 2:
    for i in range(2, (math.isqrt(n) +1)):
        if n % i == 0:
            print(f"{n}")
            break
else:
    print(f"{n} là số nguyên tố")
```

```python
# Kiểm tra số nguyên tố
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# In các số nguyên tố từ 1 đến n
n = int(input("Nhập n: "))
for x in range(1, n + 1):
    if is_prime(x):
        print(x)
```

```python
# In ra Số chia hết cho 5
n = int(input())
i = 5   #(variable) i: int
while (i <= n):
    print(i,end = ' ')
    i += 5
```

```python
# CHẠY ĐẾN SÔ n, GẶP SỐ NÀO CHIA HẾT CHO 5 THÌ DỪNG LUÔN, KHÔNG THÌ IN RA CHỮ HH
n = int(input())
i = 5
while (i <= n):
    print(i, end = ' ')       
    i += 5
    if (i == 10):
        break       # i chạy đến 10 thì dừng, đến số nào chia hết cho 5 thì dừng
else:
    print("Hoàn hảo!")
```

```python
# CỘNG DỒN TỔNG
n = int(input())
tong = 0
for i in range(n):  #cộng dồn
    tong += i
print(tong)
```

```python
# CỘNG DỒN TỔNG, MỖI SỐ CÁCH NHAU 1 ĐƠN VỊ KHI CỘNG
n = int(input())
tong = 0   # biến dùng để chứa kết quả cộng dồn, "thùng rỗng"
i = 1    #bắt đầu đếm từ 1, tức là cộng từ 1
while i <= n:  # đếm i đến khi bằng sô n thì dừng
    tong += i    # tổng cộng dồn thêm số i, rồi nhập lại vào tổng
    i += 1   # tăng biến đếm i thêm 1 đơn vị sau mỗi vòng lặp để chuyển sang số tiếp theo
print(tong)
```
