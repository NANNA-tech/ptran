```python
class XeMay:
    # 1. Khởi tạo: Thiết lập đặc điểm (Màu, Hãng)
    def __init__(self, mau_sac, hang_xe):
        self.mau = mau_sac    # Lưu màu cam vào xe
        self.hang = hang_xe   # Lưu hãng Honda vào xe

    # 2. Hành động: Chạy
    def chay(self):
        print(f"Chiếc xe {self.hang} màu {self.mau} đang chạy... Vù vù!")

    # 3. Hành động: Phanh
    def phanh(self):
        print("Két! Xe đang dừng lại.")

# --- SỬ DỤNG (Tạo ra chiếc xe thật) ---

# Tạo ra chiếc xe như bạn mô tả: Màu cam, Hãng Honda
xe_cua_ban = XeMay("Cam", "Honda")

# Sai bảo xe thực hiện hành động
xe_cua_ban.chay()   # Kết quả: Chiếc xe Honda màu Cam đang chạy... Vù vù!
xe_cua_ban.phanh()  # Kết quả: Két! Xe đang dừng lại.
```
